```python
crops = {
    "wheat": {
        "season": "Rabi",
        "temperature": "10–25°C",
        "water": "Medium",
        "soil": "Loamy soil",
        "duration": "120–150 days",
        "sunlight": "Full sunlight"
    },

    "rice": {
        "season": "Kharif",
        "temperature": "20–35°C",
        "water": "High",
        "soil": "Clayey soil",
        "duration": "120–150 days",
        "sunlight": "Full sunlight"
    },

    "maize": {
        "season": "Kharif",
        "temperature": "18–27°C",
        "water": "Medium",
        "soil": "Well-drained soil",
        "duration": "90–120 days",
        "sunlight": "Full sunlight"
    }
}


crop = input("Enter crop name: ").lower()

if crop in crops:

    information = crops[crop]

    print("\n🌾 CROP INFORMATION")
    print("----------------------")

    print("Crop:", crop.title())
    print("🌱 Season:", information["season"])
    print("🌡️ Temperature:", information["temperature"])
    print("💧 Water Requirement:", information["water"])
    print("🌍 Soil:", information["soil"])
    print("⏱️ Duration:", information["duration"])
    print("☀️ Sunlight:", information["sunlight"])

else:
    print("❌ Crop not found.")
```
