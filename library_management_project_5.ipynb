```python


books = {
    "Python": True,
    "Java": True,
    "C++": True,
    "Database": True
}

while True:
    print("\n=== LIBRARY MANAGEMENT ===")
    print("1. View Books")
    print("2. Issue Book")
    print("3. Return Book")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        for book, available in books.items():
            status = "Available" if available else "Issued"
            print(book, "-", status)

    elif choice == "2":
        book = input("Enter book name: ")

        if book in books:
            if books[book]:
                books[book] = False
                print("Book issued.")
            else:
                print("Book already issued.")
        else:
            print("Book not found.")

    elif choice == "3":
        book = input("Enter book name: ")

        if book in books:
            books[book] = True
            print("Book returned.")
        else:
            print("Book not found.")

    elif choice == "4":
        break

    else:
        print("Invalid choice.")
```
