```python
# Group Bill Splitter

print("=== Group Bill Splitter ===")

people = input("Enter names separated by commas: ").split(",")
people = [name.strip() for name in people]

payments = {}

# Enter how much each person paid
for person in people:
    amount = float(input(f"How much did {person} pay? ₹"))
    payments[person] = amount

total = sum(payments.values())
number_of_people = len(people)

# Equal share for everyone
share = total / number_of_people

print("\n========== BILL SUMMARY ==========")
print(f"Total Bill: ₹{total:.2f}")
print(f"Each Person Should Pay: ₹{share:.2f}")

# Calculate balances
balances = {}

for person in people:
    balances[person] = payments[person] - share

print("\n========== BALANCES ==========")

for person, balance in balances.items():
    if balance > 0:
        print(f"{person} should receive ₹{balance:.2f}")
    elif balance < 0:
        print(f"{person} should pay ₹{-balance:.2f}")
    else:
        print(f"{person} is settled.")

# Find who should pay whom
print("\n========== SETTLEMENT ==========")

debtors = []
creditors = []

for person, balance in balances.items():
    if balance < 0:
        debtors.append([person, -balance])
    elif balance > 0:
        creditors.append([person, balance])

i = 0
j = 0

while i < len(debtors) and j < len(creditors):

    debtor = debtors[i]
    creditor = creditors[j]

    amount = min(debtor[1], creditor[1])

    print(f"{debtor[0]} pays {creditor[0]} ₹{amount:.2f}")

    debtor[1] -= amount
    creditor[1] -= amount

    if debtor[1] == 0:
        i += 1

    if creditor[1] == 0:
        j += 1
```
